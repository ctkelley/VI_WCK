This notebook runs the examples from the paper. We do the computations from a fixed point map view point. So if $\Omega$
is the reasible region and $\bf F$ is the operator, we seek the fixed point of the map
$$
{\bf G}(x) = x - \Pi_\Omega(x - \tau {\bf F}(x))
$$
where $\Pi_\Omega$ is the projection onto $\Omega$. In each of the examples we will specify $\bf F$, $\tau$, $\Omega$, and an initial
iteration $x_0$.

With this in mind, Alg 1 is simply a direct fixed point iteration for $\bf G$. We will compare that iteration Anderson Acceleration (AA)
using the implementation from the Julia package [SIAMFANLEquations.jl](https://github.com/ctkelley/SIAMFANLEquations.jl).

We begin with Algorthm 1 and example 2 from the paper. The data are $\tau = .1$ and $x_0 = (1, 2, -1)$. We show results
from projected AA(m) for $m=0, 1, 2, 3, 10$. AA(0) is the same as Alg 1 in the paper. I set a maximum iteration limit of $100$ and the $m=0$
case had not converged yet, hence the failure message.

The code is a projected version of the AA code from my book with changes to account for the bounds.
